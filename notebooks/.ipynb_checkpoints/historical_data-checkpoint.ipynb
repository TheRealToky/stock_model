# Functions

```python
def get_balancesheet(ticker_shadows):
    tick = yf.Ticker(ticker_shadows)

    tick_bs = tick.balancesheet
    df_bs = pd.DataFrame(tick_bs)
    return df_bs


def get_historical_data(ticker_shadows):
    tick = yf.Ticker(ticker_shadows)

    historical_data = tick.history(period='max', interval='1d')
    df_historical_data = pd.DataFrame(historical_data)
    return df_historical_data


def get_information (ticker_shadows):
    tick = yf.Ticker(ticker_shadows)

    tick_info = tick.info  # ticker_shadows's general info
    # tick_info = json.dumps(tick_info, indent=4)
    return tick_info


def get_financials (ticker_shadows):
    tick = yf.Ticker(ticker_shadows)

    financials = tick.financials
    df_financials = pd.DataFrame(financials)
    return df_financials


def get_cashflow (ticker_shadows):
    tick = yf.Ticker(ticker_shadows)

    tick_cashflow = tick.cashflow  # shows the cashflow data
    df_tick_cashflow = pd.DataFrame(tick_cashflow)
    return df_tick_cashflow
```
