# Imports

This file fetches all data from individual tickers and adds them to the database.

```python
import pandas as pd
import yfinance as yf
```

# Functions



```python
def get_balancesheet(ticker_symbol):
    try:
        tick = yf.Ticker(ticker_symbol)

        tick_bs = tick.balancesheet
        df_bs = pd.DataFrame(tick_bs)
        return df_bs
    except Exception as e:
        print(f"[!] Failed to fetch balancesheet {ticker_symbol}:")
        print(f"[!] {e}")


def get_historical_data(ticker_symbol, period="1d", interval="1m"):
    try:
        tick = yf.Ticker(ticker_symbol)

        raw_historical_data = tick.history(period=period, interval=interval)
        df_historical_data = pd.DataFrame(raw_historical_data)
        return df_historical_data
    except Exception as e:
        print(f"[!] Failed to fetch historical data {ticker_symbol}:")
        print(f"[!] {e}")

def get_information(ticker_symbol):
    try:
        tick = yf.Ticker(ticker_symbol)

        tick_info = tick.info  # ticker_symbol's general info
        # tick_info = json.dumps(tick_info, indent=4)
        return tick_info
    except Exception as e:
        print(f"[!] Failed to fetch company information {ticker_symbol}:")
        print(f"[!] {e}")

def get_financials(ticker_symbol):
    try:
        tick = yf.Ticker(ticker_symbol)

        raw_financials = tick.financials
        df_financials = pd.DataFrame(raw_financials)
        return df_financials
    except Exception as e:
        print(f"[!] Failed to fetch financials {ticker_symbol}:")
        print(f"[!] {e}")

def get_cashflow(ticker_symbol):
    try:
        tick = yf.Ticker(ticker_symbol)

        tick_cashflow = tick.cashflow  # shows the cashflow data
        df_tick_cashflow = pd.DataFrame(tick_cashflow)
        return df_tick_cashflow
    except Exception as e:
        print(f"[!] Failed to fetch cashflow {ticker_symbol}:")
        print(f"[!] {e}")
def get_all(ticker_symbol):
    try:
        historical_data = get_historical_data(ticker_symbol)
        information =  get_information(ticker_symbol)
        financials = get_financials(ticker_symbol)
        cashflow = get_cashflow(ticker_symbol)
        balancesheet = get_balancesheet(ticker_symbol)

        return historical_data, information, financials, cashflow, balancesheet
    except Exception as e:
        print(f"[!] Failed to fetch all {ticker_symbol}:")
        print(f"[!] {e}")
```

## Download the top 50 stocks

```python
import os
import time

for symbol in stock_by_market_cap["Symbol"].head(50):
    file_path = f"../data/historical_data/{symbol}.csv"
    failed_list = []

    if os.path.isfile(file_path):
        print(f"{symbol} already exists.")
        continue
    else:
        print(f"Downloading {symbol} data...")

        try:
            symbol_historical_data, symbol_information, symbol_financials, symbol_cashflow, symbol_balancesheet = get_all(symbol)

            symbol_historical_data.to_csv(f'../data/historical_data/{symbol}.csv')
            symbol_financials.to_csv(f'../data/stock_financials/{symbol}.csv')
            symbol_cashflow.to_csv(f'../data/stock_cashflow/{symbol}.csv')
            symbol_balancesheet.to_csv(f'../data/stock_balancesheet/{symbol}.csv')

            # Turns out, dictionaries can't be converted to csv XD
            # symbol_information.to_csv(f'../data/stock_info/{symbol}.csv')

            time.sleep(10)
        except Exception as error:
            print(f"Failed to download {symbol}:")
            print(error)
            failed_list.append(symbol)
```
